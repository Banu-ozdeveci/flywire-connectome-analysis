### 1. What “community size” means in practice

When we say community size, we mean:

- how many nodes end up grouped together

- how coarse or fine the partition is

Different algorithms optimize different objectives, so they naturally prefer different sizes.

### 2. Louvain: size is driven by modularity optimization

What Louvain optimizes

Standard (undirected) modularity

##### “Are there more edges inside groups than expected by chance?”

Consequences

Louvain has a well-known resolution limit

It tends to:

- merge small but meaningful groups

- favor fewer, larger communities

In our  result

- 1913 nodes

- 6 communities

This is very typical for Louvain on dense biological networks.

### 3. Biomodularity: size emerges from directional roles

What biomodularity optimizes

- Deviations from expected sending vs receiving patterns

- Uses a directed null model

Works in a low-rank latent space (SVD-based)

##### Consequences

Communities are defined by roles, not just density

We can have:

- nodes in the same sending community but different receiving communities

- smaller, more functionally specific groups

##### Typical outcome

- More communities than Louvain

- More uneven community sizes

- Some communities can be small but still meaningful


Biomodularity answers:

“Which neurons behave similarly as senders or receivers?”
