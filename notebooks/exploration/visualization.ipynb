```python
import pandas as pd
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import seaborn as sns
import pickle

plt.style.use("seaborn-v0_8")

with open("../data/preprocessed/subgraph.pkl", "rb") as f:
    H = pickle.load(f)

print("Loaded H:", H.number_of_nodes(), "nodes,", H.number_of_edges(), "edges")

# Load centralities
cent_df = pd.read_csv("../data/preprocessed/node_centralities.csv")
cent_df.head()
```

```python
# Compute Louvain communities and store on nodes
from networkx.algorithms.community import louvain_communities

communities = louvain_communities(H, weight='weight', seed=42)
node_comm = {}
for cid, nodes in enumerate(communities):
    for n in nodes:
        node_comm[n] = cid
nx.set_node_attributes(H, node_comm, 'community')

sizes = [len(c) for c in communities]
print(f'Detected {len(communities)} communities; largest size={max(sizes) if sizes else 0}')
```

```python
# Degree distribution
deg = [d for _, d in H.degree()]

plt.figure(figsize=(6,5))
plt.hist(deg, bins=40, color="steelblue", alpha=0.7)
plt.xscale("log")
plt.yscale("log")
plt.xlabel("Degree")
plt.ylabel("Count")
plt.title("Degree Distribution (Log-Log Scale)")
plt.tight_layout()
plt.show()
```

```python
#Degree vs Centrality
plt.figure(figsize=(6,5))
sns.scatterplot(
    data=cent_df,
    x="degree_centrality",
    y="betweenness",
    alpha=0.6,
    color="mediumseagreen"
)
plt.xscale("log")
plt.yscale("log")
plt.title("Degree Centrality vs Betweenness")
plt.tight_layout()
plt.show()
```

```python
#Page rank vs Degree
plt.figure(figsize=(6,5))
sns.scatterplot(
    data=cent_df,
    x="degree_centrality",
    y="pagerank",
    alpha=0.6,
    color="crimson"
)
plt.xscale("log")
plt.yscale("log")
plt.title("Degree Centrality vs PageRank")
plt.tight_layout()
plt.show()

```

```python
# Get community ID for each node
comms = [H.nodes[n].get('community', 0) for n in H.nodes()]
pos = nx.spring_layout(H, seed=42, k=0.15)

plt.figure(figsize=(8,7))
nx.draw_networkx_nodes(
    H, pos,
    node_color=comms,
    cmap='tab20',
    node_size=25,
    alpha=0.9
 )
nx.draw_networkx_edges(H, pos, alpha=0.05)
plt.title('Louvain Communities (Subgraph)')
plt.axis('off')
plt.show()
```
